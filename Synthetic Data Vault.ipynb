```python
import pandas as pd
from sdv.single_table import GaussianCopulaSynthesizer, CTGANSynthesizer
from sdv.evaluation.single_table import run_diagnostic, evaluate_quality, get_column_plot, get_column_pair_plot
from sdv.metadata import SingleTableMetadata, Metadata
```

```python
df = pd.read_csv('Original Data Clean.csv')
metadata = Metadata.detect_from_dataframe(data = df)
```

```python
synthesizer = CTGANSynthesizer(metadata, epochs = 500, verbose = True, enforce_rounding = False)
```

```python
synthesizer.fit(df)
synthetic_data = synthesizer.sample(num_rows = (5000 - len(df)))
```

```python
quality = evaluate_quality(df, synthetic_data, metadata)
```

```python
fig = get_column_plot(real_data = df, synthetic_data = synthetic_data, column_name = 'Power Generated', metadata = metadata)
fig.show()
```

```python
from sdv.evaluation.single_table import run_diagnostic

diagnostic = run_diagnostic(
    real_data=df,
    synthetic_data=synthetic_data,
    metadata=metadata
)
```
